```python
# write lines
L = ['hello\n','hi\n','how are you\n','I am fine']

f = open('/home/sample.txt','w')
f.writelines(L)
f.close()
```

```python
f = open('/home/sample.txt','r')
s = f.read()
print(s)
f.close()
```

```python
# reading upto n chars
f = open('/home/sample.txt','r')
s = f.read(15)
print(s)
f.close()
```

```python
# readline() -> to read line by line
f = open('/home/sample.txt','r')
print(f.readline(),end='')
print(f.readline(),end='')
f.close()
```

```python
# reading entire using readline
f = open('/home/sample.txt','r')

while True:
    data = f.readline()

    if data == "":
        break
    else:
        print(data, end='')

f.close()
```

```python
# moving within a file
with open("/home/sample.txt", "r") as f:
    print(f.read(10))
    print(f.read(10))
    print(f.read(10))
```

```python
# benefit? -> to load a big file in memory
big_L = ['hello world ' for i in range(1000)]

with open('/home/big.txt','w') as f:
  f.writelines(big_L)
```

```python
with open("/home/big.txt",'r') as f:
    chunk_size = 10

    while len(f.read(chunk_size)) > 0:
        print(f.read(chunk_size),end="")
        f.read(chunk_size)
```

```python
# seek and tell function

with open("/home/sample.txt", "r") as f:
    print(f.read(10))
    print(f.read(10))
    print(f.tell())
```

```python
with open("/home/sample.txt", "r") as f:

    print(f.read(10))
    f.seek(0)
    print(f.tell())
```

```python
with open("/home/sample.txt", "w") as f:
    f.write("Hello")
    f.seek(0)
    f.write("Da")
    print(f.tell())
```
