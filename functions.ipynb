```python
def multiply(*args):
    product = 1
    for i in args:
        product = product * i

    return product
res = multiply(1,2,3,3,4,5,6,5,6,7,45)
print(res)
```

```python
def unique_items(L):
    results = []
    for i in L:
        if i not in results:
            results.append(i)
    return results

L = [1,2,3,3,3,3,4,5]
out = unique_items(L)
print(out)
```
