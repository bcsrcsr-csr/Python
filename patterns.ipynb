```python
for i in range(5):
    for j in range(5):
        print("*", end=" ")
    print()
```

```python
for i in range(1, 6):
    for j in range(i):
        print("*", end=" ")
    print()
```

```python
for i in range(5, 0, -1):
    for j in range(i):
        print("*", end=" ")
    print()
```

```python
for i in range(1, 6):
    for j in range(5 - i):
        print(" ", end=" ")
    for k in range(i):
        print("*", end=" ")
    print()
```

```python
for i in range(1, 6):
    for j in range(5 - i):
        print(" ", end=" ")
    for k in range(2 * i - 1):
        print("*", end=" ")
    print()
```

```python
for i in range(5, 0, -1):
    for j in range(5 - i):
        print(" ", end=" ")
    for k in range(2 * i - 1):
        print("*", end=" ")
    print()
```

```python
n = 5

for i in range(1, n + 1):
    print(" " * (n - i) + "*" * (2 * i - 1))

for i in range(n - 1, 0, -1):
    print(" " * (n - i) + "*" * (2 * i - 1))
```

```python
n = 5

for i in range(n):
    for j in range(n):
        if i == 0 or i == n-1 or j == 0 or j == n-1:
            print("*", end=" ")
        else:
            print(" ", end=" ")
    print()
```

```python
n = 5

for i in range(1, n + 1):
    for j in range(1, i + 1):
        if j == 1 or j == i or i == n:
            print("*", end=" ")
        else:
            print(" ", end=" ")
    print()
```

```python
for i in range(1, 6):
    for j in range(i):
        print(i, end=" ")
    print()
```

```python
num = 1

for i in range(1, 6):
    for j in range(i):
        print(num, end=" ")
        num += 1
    print()
```

```python
for i in range(5, 0, -1):
    for j in range(1, i + 1):
        print(j, end=" ")
    print()
```

```python
for i in range(1, 6):
    for j in range(i):
        print((i + j + 1) % 2, end=" ")
    print()
```

```python
for i in range(1, 6):
    for j in range(i):
        print(chr(65 + j), end=" ")
    print()
```

```python
for i in range(1, 6):
    for j in range(i):
        print(chr(64+i), end=" ")
    print()
```

```python
for i in range(1, 6):
    print("  " * (5-i), end=" ")

    for j in range(i):
        print(chr(65+j), end=" ")

    for j in range(i-2, -1, -1):
        print(chr(65+j), end=" ")

    print()
```
