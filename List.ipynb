```python
cars_name = ["Audi", "BMW", "TATA", "Volvo"]
no_of_cars = [10,20,30,5]

for i,j in zip(cars_name, no_of_cars):
    print(f"{i} = {j}")
```

```python
LC = [f"{i} = {j}" for i,j in zip(cars_name, no_of_cars)]
```

```python
L = [2,4,6,10,1]

result = []

for i in L:
    sum = 0
    for j in L:
        if i<=j:
            sum = sum + j
    result.append(sum)


print(result)
```

```python
num1 = [23,45,67,78,89,34,67]
num2 = [34,89,55,56,39,67,67]

common = []

for i in num1:
    if i in num2:
        if i not in common:
            common.append(i)



print(common)
```
