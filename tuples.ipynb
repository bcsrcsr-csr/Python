```python
v = [2,3,4]
s = -3

L = []

for i in v:
    result = i * s
    L.append(result)

print(L)
```

```python
TC = tuple((i*s for i in v))
print(TC)
```

```python
T = (1,2,3,4)

TC = tuple((i**2 for i in T))

print(TC)
```

```python
for i in range(1,51):
    if i%5==0:
        print(i)
```

```python
TC = tuple((i for i in range(1,51) if i%5==0))
print(TC)
```

```python
languages = ['java','python','php','c','javascript']

TC = tuple((language for language in languages if language.startswith("p")))
print(TC)
```
