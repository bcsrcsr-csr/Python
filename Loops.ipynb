```python
for i in range(1,11):
    print(i)
```

```python
for i in range(10,0,-1):
    print(i, end =" ")
```

```python
for i in range(1,11,3):
    print(i)
```

```python

for i in "Sasidhar":
    print(i)
```

```python
for i in [1,2,3,4]:
    print(i)
```

```python
i = 1

while i <= 5:
    print(i)
    i = i + 1
```

```python
fruits = ["Apple", "Banana", "Mango", "Orange"]

for fruit in fruits:
    print(fruit)
```

```python
for i in range(1, 6):
    if i == 3:
        break
    print(i)
```

```python
for i in range(1, 4):
    for j in range(1, 4):
        print(i, j)
```

```python
i = 1

while True:
    print(i)
    i = i + 1

    if i > 5:
        break
```
